```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import pandas as pd
import time

#configurar opções
options = Options()
options.add_argument('--headless')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')

#criar driver
driver = webdriver.Chrome(options=options)

#acessar o site
url = "http://quotes.toscrape.com"
driver.get(url)

#extrair dados
citacoes = []
autores = []
tags = []

while True:
    quote_elements = driver.find_elements(By.CLASS_NAME, "quote")
    for quote in quote_elements:
        citacoes.append(quote.find_element(By.CLASS_NAME, "text").text)
        autores.append(quote.find_element(By.CLASS_NAME, "author").text)
        tags.append(", ".join([tag.text for tag in quote.find_elements(By.CLASS_NAME, "tag")]))
    try:
        next_button = driver.find_element(By.CLASS_NAME, "next")
        next_button.click()
    except:
        break

driver.quit()

#criar DataFrame
df = pd.DataFrame({
    "Citação": citacoes,
    "Autor": autores,
    "Tags": tags
})
df.head()
```
