```python
#encontrar todos os elementos de citação na página
quotes = driver.find_elements(By.CLASS_NAME, 'quote')

#listas para armazenar os dados extraídos
citacoes = []
autores = []
tags = []

#iterar sobre os elementos e extrair as informações
for quote in quotes:
    #extrair a citação
    citacao = quote.find_element(By.CLASS_NAME, 'text').text
    citacoes.append(citacao)

    #extrair o autor
    autor = quote.find_element(By.CLASS_NAME, 'author').text
    autores.append(autor)

    #extrair as tags
    tag_elements = quote.find_elements(By.CLASS_NAME, 'tag')
    tag_text = [tag.text for tag in tag_elements]
    tags.append(tag_text)
```

```python
#criar o DataFrame com os dados extraídos
df = pd.DataFrame({
    'Citação': citacoes,
    'Autor': autores,
    'Tags': tags
})

#exibir as primeiras linhas do DataFrame
df.head()
```

```python
#navegar por várias páginas (opcional)
while True:
    #extrair os dados da página atual
    quotes = driver.find_elements(By.CLASS_NAME, 'quote')

    #armazenar os dados da página
    for quote in quotes:
        citacao = quote.find_element(By.CLASS_NAME, 'text').text
        autor = quote.find_element(By.CLASS_NAME, 'author').text
        tag_elements = quote.find_elements(By.CLASS_NAME, 'tag')
        tag_text = [tag.text for tag in tag_elements]

        citacoes.append(citacao)
        autores.append(autor)
        tags.append(tag_text)

    #tentar acessar a próxima página
    try:
        next_button = driver.find_element(By.CLASS_NAME, 'next')
        next_button.click()
        time.sleep(2)  # Esperar a página carregar
    except:
        print("Última página atingida!")
        break
```

```python
# Fechar o navegador
driver.quit()
print("Página fechada!")
```
